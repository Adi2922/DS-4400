```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_curve, auc
from sklearn.preprocessing import StandardScaler
```

```python

#this is path to the dataset and the dataset is getting loaded using pandas
path = "/content/spambase.data"
df = pd.read_csv(path, header=None)
```

```python
df
```

```python

#these are all the columns from the spambase .names file
feat = ['word_freq_make', 'word_freq_address', 'word_freq_all', 'word_freq_3d',
    'word_freq_our', 'word_freq_over', 'word_freq_remove', 'word_freq_internet',
    'word_freq_order', 'word_freq_mail', 'word_freq_receive', 'word_freq_will',
    'word_freq_people', 'word_freq_report', 'word_freq_addresses', 'word_freq_free',
    'word_freq_business', 'word_freq_email', 'word_freq_you', 'word_freq_credit',
    'word_freq_your', 'word_freq_font', 'word_freq_000', 'word_freq_money',
    'word_freq_hp', 'word_freq_hpl', 'word_freq_george', 'word_freq_650',
    'word_freq_lab', 'word_freq_labs', 'word_freq_telnet', 'word_freq_857',
    'word_freq_data', 'word_freq_415', 'word_freq_85', 'word_freq_technology',
    'word_freq_1999', 'word_freq_parts', 'word_freq_pm', 'word_freq_direct',
    'word_freq_cs', 'word_freq_meeting', 'word_freq_original', 'word_freq_project',
    'word_freq_re', 'word_freq_edu', 'word_freq_table', 'word_freq_conference',
    'char_freq_;', 'char_freq_(', 'char_freq_[', 'char_freq_!', 'char_freq_$',
    'char_freq_#', 'capital_run_length_average', 'capital_run_length_longest',
    'capital_run_length_total', 'spam']

df.columns = feat
# we take all features in X and drop all columns except spam for y
X = df.drop('spam', axis=1).values
y = df['spam'].values
```

```python
#1.1

# we split the data into 75 percent training and 25 percent testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

mod1 = LogisticRegression(max_iter=1000)
mod1.fit(X_train, y_train)
```

```python


y_pred = mod1.predict(X_test)
probs = mod1.predict_proba(X_test)[:,1]
```

```python

#here we compute the confusion matrix and the accuracy error precison recall and f1 score
cm = confusion_matrix(y_test, y_pred)
print("The confusion matrix is "+ str(cm))


a = accuracy_score(y_test, y_pred)
e = 1 - a
p = precision_score(y_test, y_pred)
r = recall_score(y_test, y_pred)
f = f1_score(y_test, y_pred)
print("The accuracy is " + str(a))
print("The error is "+ str(e))
print("The precison is "+ str(p))
print("The recall value is "+ str(r))
print("The f1 score is "+ str(f))
```

```python
#1.2

# Here we are analyzing feature coefficients to see which contribute most to spam prediction
feature_names = feat[:-1]
data = {'feature': feature_names, 'coef': mod1.coef_[0]}
c = pd.DataFrame(data)
c = c.sort_values('coef', ascending=False)

# Here the positive coefficients = positively correlated with spam, negative = negatively correlated
print("Top 5 positive (spam indicators):")
print(c.head(5))
print("\n")
print("Top 5 negative ")
print(c.tail(5))


```

```python
# 1.3

def calculateBasedOnThreshold(probs, y_true, t):
    tp = 0
    fp = 0
    fn = 0
    tn = 0
    for i in range(len(y_true)):
        if probs[i] >= t:
          pred = 1
        else:
          pred = 0


        #if predicted and y true is same then it is true positive
        if pred == 1 and y_true[i] == 1:
            tp += 1
        #if predicted is 1 and y true is 0 then it is a false positive
        elif pred == 1 and y_true[i] == 0:
            fp += 1
        #if predicted is 0 and y true is 1 then it is a false negative
        elif pred == 0 and y_true[i] == 1:
            fn += 1
        else:
          #if predicted is 0 and y true is 0 then it is a true negative
            tn += 1

    a = (tp + tn) / len(y_true)

    if (tp + fp) > 0:
      p = tp / (tp + fp)
    else:
      p = 0

    if (tp + fn) > 0:
      r = tp / (tp + fn)
    else:
      r = 0
    return a, p, r


for i in [0.25, 0.5, 0.75, 0.9]:
    acc, prec, rec = calculateBasedOnThreshold(probs, y_test, i)
    print("Threshold=" + str(i))
    print ("accuracy=" + str(acc) + " precision=" + str(prec) + " recall=" + str(rec))
    print()
```
